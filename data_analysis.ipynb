```python
df["cost_price_ratio"] = df["cost"] / df["price"]
df["gross_profit_per_unit"] = df["price"] - df["cost"]
df["estimated_profit"] = df["gross_profit_per_unit"] * df["sales"]
```

```python
profit_by_category = df.groupby("category").agg(
    avg_profit_margin=("profit_margin", "mean"),
    avg_cost_price_ratio=("cost_price_ratio", "mean"),
    total_estimated_profit=("estimated_profit", "sum"),
    avg_estimated_profit=("estimated_profit", "mean")
).sort_values(by="total_estimated_profit", ascending=False)

display(profit_by_category)
```

```python
df["inventory_sales_ratio"] = df["inventory"] / (df["sales"] + 1)
df["sales_inventory_ratio"] = df["sales"] / (df["inventory"] + 1)
```

```python
inventory_by_category = df.groupby("category").agg(
    total_inventory=("inventory", "sum"),
    total_sales=("sales", "sum"),
    avg_inventory=("inventory", "mean"),
    avg_sales=("sales", "mean"),
    avg_inventory_sales_ratio=("inventory_sales_ratio", "mean")
).sort_values(by="avg_inventory_sales_ratio", ascending=False)

display(inventory_by_category)
```

```python
sales_high = df["sales"].quantile(0.75)
sales_low = df["sales"].quantile(0.25)
inventory_high = df["inventory"].quantile(0.75)
inventory_low = df["inventory"].quantile(0.25)

def classify_inventory_status(row):
    if row["inventory"] >= inventory_high and row["sales"] <= sales_low:
        return "Overstock Risk"
    elif row["inventory"] <= inventory_low and row["sales"] >= sales_high:
        return "Stockout Risk"
    else:
        return "Normal"

df["inventory_risk"] = df.apply(classify_inventory_status, axis=1)

inventory_risk_summary = df["inventory_risk"].value_counts()

display(inventory_risk_summary)
```

```python
overstock_products = df[df["inventory_risk"] == "Overstock Risk"].sort_values(
    by="inventory_sales_ratio",
    ascending=False
)

display(overstock_products[
    ["category", "brand", "material", "location", "store_type", "inventory", "sales", "inventory_sales_ratio", "profit_margin"]
].head(10))
```

```python
stockout_products = df[df["inventory_risk"] == "Stockout Risk"].sort_values(
    by="sales_inventory_ratio",
    ascending=False
)

display(stockout_products[
    ["category", "brand", "material", "location", "store_type", "inventory", "sales", "sales_inventory_ratio", "profit_margin"]
].head(10))
```

```python
df["delay_risk"] = np.where(df["delivery_days"] > 7, 1, 0)

delay_risk_summary = df.groupby("delay_risk").agg(
    count=("delay_risk", "count"),
    avg_delivery_days=("delivery_days", "mean")
)

display(delay_risk_summary)
```

```python
delay_by_location = df.groupby("location").agg(
    delay_rate=("delay_risk", "mean"),
    avg_delivery_days=("delivery_days", "mean")
).sort_values(by="delay_rate", ascending=False)

display(delay_by_location)
```
