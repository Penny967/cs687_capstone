```python
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)

print("Dataset shape:")
print(df.shape)

print("Columns:")
print(df.columns.tolist())

display(df.head())
```

```python
category_sales_revenue = df.groupby("category").agg(
    total_sales=("sales", "sum"),
    total_revenue=("revenue", "sum"),
    avg_sales=("sales", "mean"),
    avg_revenue=("revenue", "mean")
).sort_values(by="total_revenue", ascending=False)

display(category_sales_revenue)
```

```python
category_sales_revenue["total_sales"].plot(kind="bar", figsize=(8, 5))
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
brand_sales_revenue = df.groupby("brand").agg(
    total_sales=("sales", "sum"),
    total_revenue=("revenue", "sum"),
    avg_sales=("sales", "mean"),
    avg_revenue=("revenue", "mean")
).sort_values(by="total_revenue", ascending=False)

display(brand_sales_revenue)
```

```python
category_sales_revenue["total_revenue"].plot(kind="bar", figsize=(8, 5))
plt.title("Total Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
material_sales_revenue = df.groupby("material").agg(
    total_sales=("sales", "sum"),
    total_revenue=("revenue", "sum"),
    avg_sales=("sales", "mean"),
    avg_revenue=("revenue", "mean")
).sort_values(by="total_revenue", ascending=False)

display(material_sales_revenue)
```

```python
category_sales_revenue["total_revenue"].plot(kind="bar", figsize=(8, 5))
plt.title("Total Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
season_analysis = df.groupby("season").agg(
    total_sales=("sales", "sum"),
    total_revenue=("revenue", "sum"),
    avg_sales=("sales", "mean"),
    avg_revenue=("revenue", "mean")
).sort_values(by="total_sales", ascending=False)

display(season_analysis)
```

```python
location_analysis = df.groupby("location").agg(
    total_sales=("sales", "sum"),
    total_revenue=("revenue", "sum"),
    avg_sales=("sales", "mean"),
    avg_revenue=("revenue", "mean")
).sort_values(by="total_sales", ascending=False)

display(location_analysis)
```

```python
season_category_sales = pd.pivot_table(
    df,
    values="sales",
    index="season",
    columns="category",
    aggfunc="sum"
)

display(season_category_sales)
```

```python
location_category_sales = pd.pivot_table(
    df,
    values="sales",
    index="location",
    columns="category",
    aggfunc="sum"
)

display(location_category_sales)
```

```python
season_category_sales.plot(kind="bar", figsize=(10, 6))
plt.title("Sales by Season and Category")
plt.xlabel("Season")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.show()
```

```python
location_category_sales.plot(kind="bar", figsize=(10, 6))
plt.title("Sales by Location and Category")
plt.xlabel("Location")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.show()
```

```python
df["cost_price_ratio"] = df["cost"] / df["price"]
df["gross_profit_per_unit"] = df["price"] - df["cost"]
df["estimated_profit"] = df["gross_profit_per_unit"] * df["sales"]
```

```python
profit_by_category = df.groupby("category").agg(
    avg_profit_margin=("profit_margin", "mean"),
    avg_cost_price_ratio=("cost_price_ratio", "mean"),
    total_estimated_profit=("estimated_profit", "sum"),
    avg_estimated_profit=("estimated_profit", "mean")
).sort_values(by="total_estimated_profit", ascending=False)

display(profit_by_category)
```

```python
profit_by_material = df.groupby("material").agg(
    avg_profit_margin=("profit_margin", "mean"),
    avg_cost_price_ratio=("cost_price_ratio", "mean"),
    total_estimated_profit=("estimated_profit", "sum"),
    avg_estimated_profit=("estimated_profit", "mean")
).sort_values(by="total_estimated_profit", ascending=False)

display(profit_by_material)
```

```python
high_profit_products = df.sort_values(by="profit_margin", ascending=False).head(10)

display(high_profit_products[
    ["category", "brand", "material", "color", "store_type", "location", "price", "cost", "profit_margin", "sales", "revenue"]
])
```

```python
high_revenue_low_profit = df[
    (df["revenue"] >= df["revenue"].quantile(0.75)) &
    (df["profit_margin"] <= df["profit_margin"].quantile(0.25))
]

display(high_revenue_low_profit[
    ["category", "brand", "material", "price", "cost", "sales", "revenue", "profit_margin"]
].sort_values(by="revenue", ascending=False).head(10))
```
